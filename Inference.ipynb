## **Inference/Testing Model**

Testing yang menghasilkan output berupa kelas kategorikal (contoh: negatif, netral, dan positif).

**Import Library**

```python
import numpy as np
import tensorflow as tf
import pickle
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

**Load Model**

```python
model_gru = tf.keras.models.load_model("Model/gru_model_brimo.h5")
model_lstm = tf.keras.models.load_model("Model/lstm_model.h5")
model_cnn = tf.keras.models.load_model("Model/cnn_model.h5")

# Muat tokenizer dan label encoder
with open("Model/tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

with open("Model/label_encoder.pkl", "rb") as f:
    label_encoder = pickle.load(f)
```

**Input dan Prediksi**

```python
# Contoh input teks baru yang ingin diprediksi
input_texts = [
    "Aplikasinya bagus dan sangat mudah digunakan.",
    "Tiap buka aplikasi selalu force close, capek.",
    "Sudah oke tapi kadang agak lambat responnya."
]

# Tokenisasi dan padding
seqs = tokenizer.texts_to_sequences(input_texts)
padded = pad_sequences(seqs, maxlen=50)

# Prediksi dari ketiga model
preds_gru = model_gru.predict(padded)
preds_lstm = model_lstm.predict(padded)
preds_cnn = model_cnn.predict(padded)
```

**Interpretasi dan Tampilkan Hasil**

```python
# Ubah hasil prediksi ke label sentimen
label_gru = label_encoder.inverse_transform(np.argmax(preds_gru, axis=1))
label_lstm = label_encoder.inverse_transform(np.argmax(preds_lstm, axis=1))
label_cnn = label_encoder.inverse_transform(np.argmax(preds_cnn, axis=1))

# Tampilkan hasil
print("\n=== Hasil Prediksi Sentimen ===\n")
for i, text in enumerate(input_texts):
    print(f"Teks: {text}")
    print(f"Prediksi GRU:  {label_gru[i]}")
    print(f"Prediksi LSTM: {label_lstm[i]}")
    print(f"Prediksi CNN:  {label_cnn[i]}")
    print("-" * 60)
```
