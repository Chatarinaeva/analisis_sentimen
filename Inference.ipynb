## **Inference/Testing Model**

Testing yang menghasilkan output berupa kelas kategorikal (contoh: negatif, netral, dan positif).

**Import Library**

```python
import numpy as np
import tensorflow as tf
import pickle
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

**Load Model** 

```python
gru_model = tf.keras.models.load_model("Model/gru_model_brimo.h5")
lstm_model = tf.keras.models.load_model("Model/lstm_model.h5")
cnn_model = tf.keras.models.load_model("Model/cnn_model.h5")

# Muat tokenizer dan label encoder
with open("Model/tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

with open("Model/label_encoder.pkl", "rb") as f:
    label_encoder = pickle.load(f)
```

**Testing**

```python
sample_texts = [
    "sangat puas dan nyaman digunakan",
    "aplikasi selalu error",
    "praktis meskipun kadang muncul error."
]

# Label asli sebagai pembanding
true_sentiments = ["positive", "negative", "neutral"]

# Tokenisasi dan padding
sample_sequences = tokenizer.texts_to_sequences(sample_texts)
sample_inputs = pad_sequences(sample_sequences, maxlen=50)

# Prediksi dengan ketiga model
proba_gru  = gru_model.predict(sample_inputs)
proba_lstm = lstm_model.predict(sample_inputs)
proba_cnn  = cnn_model.predict(sample_inputs)

# Konversi probabilitas ke label indeks
pred_gru  = np.argmax(proba_gru, axis=1)
pred_lstm = np.argmax(proba_lstm, axis=1)
pred_cnn  = np.argmax(proba_cnn, axis=1)

# Dekode label dari angka ke string
label_gru  = label_encoder.inverse_transform(pred_gru)
label_lstm = label_encoder.inverse_transform(pred_lstm)
label_cnn  = label_encoder.inverse_transform(pred_cnn)

# Tampilkan hasil prediksi per model
for text, true, gru_pred, lstm_pred, cnn_pred in zip(sample_texts, true_sentiments, label_gru, label_lstm, label_cnn):
    print("-" * 60)
    print(f"Teks: {text}")
    print(f"Label Asli: {true}")
    print(f"Prediksi GRU:  {gru_pred}")
    print(f"Prediksi LSTM: {lstm_pred}")
    print(f"Prediksi CNN:  {cnn_pred}")
print("-" * 60)
```
